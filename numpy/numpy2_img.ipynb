## Image manipulation with numpy

# 1. Using convolution kernal for edge detection 
This was heavily inspired by 3 blue 1 brown video convolve

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.ndimage import convolve

# Load the image
photo1 = plt.imread('./img1.jpg')
# Convert to grayscale using weighted average
grey = np.dot(photo1[..., :3], [0.299, 0.3, 0.3])

# Define the convolution kernel (e.g., edge detection or similar)
converter = np.array([
    [0.25, 0, -0.25],
    [0.5, 0, -0.5],
    [0.25, 0, -0.25]
])

# Apply convolution
convol = convolve(grey, converter)
convol=np.where(convol<55,convol*255,0)
# Display the result
plt.imshow(convol, cmap='gray')
plt.axis('off')
plt.title("Convolved Image")
plt.show()
```

# 2. Showing the pixel data using numpy

```python
#Each pixel of an image contains 3 values ranging from 0-255, 
x= np.array(photo1)
print(x)
# np.shape gives the image values of the height, width and 3(for jpeg there are always 3 values=> r,g and b 
# for png extra alpha chanel ranging from 0-100 is also present)
print(x.shape)
```

# 3. Extracting Only one channel

```python
# Convert to numpy array, then extract the channel, and use extra cmap functions
photo1_np=np.array(photo1)
red_channel=photo1_np[:,:,0]
plt.title("Only red")
plt.imshow(red_channel,cmap="Reds")
```
