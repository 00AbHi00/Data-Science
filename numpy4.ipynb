```python
import pandas as pd
import matplotlib.pyplot as plt

# Create a simple DataFrame
data = pd.DataFrame(
    [
        ['Abhishek', 16],
        ['Nokia', 10]
    ],
    columns=['Name', 'Age']
)

# Display the original DataFrame
print("Original Data:")
print(data)

# 1. Unique names
print("\nUnique Names:")
print(data['Name'].unique())

# 2. Frequency of each name
print("\nName Counts:")
print(data['Name'].value_counts())

# 3. Length of each name
print("\nLength of Each Name:")
print(data['Name'].str.len())

# 4. First and last character of each name
print("\nFirst Character of Each Name:")
print(data['Name'].str[0])

print("\nLast Character of Each Name:")
print(data['Name'].str[-1])

# 5. Most common starting letter
print("\nMost Common Starting Letter:")
print(data['Name'].str[0].value_counts())

# 6. Check if the name contains the letter 'a' (case-insensitive)
print("\nDoes the Name Contain 'a'? (case-insensitive):")
print(data['Name'].str.contains('a', case=False))

# 7. Convert names to uppercase
print("\nNames in Uppercase:")
print(data['Name'].str.upper())

# 8. Grouping: average age by name
print("\nAverage Age by Name:")
print(data.groupby('Name')['Age'].mean())

# 9. Optional: Simple name frequency bar chart using matplotlib
name_counts = data['Name'].value_counts()
name_counts.plot(kind='bar', title="Name Frequency")
plt.xlabel("Name")
plt.ylabel("Count")
plt.show()
```
