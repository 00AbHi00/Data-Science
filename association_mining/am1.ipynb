## Association Mining 
Association mining is the process of finding frequent itemsets (groups of items that occur together often) and deriving association rules that highlight relationships between items in transactional data.

```python
import pandas as pd
# Step 1: Load the dataset
data = {
'TransactionID': [1, 2, 3, 4, 5],
'Items': [
['Bread', 'Milk'],
['Bread', 'Diaper', 'Beer', 'Eggs'],
['Milk', 'Diaper', 'Beer', 'Coke'],
['Bread', 'Milk', 'Diaper', 'Beer'],
['Bread', 'Milk', 'Diaper', 'Coke']
]
}
df = pd.DataFrame(data)
print("Initial Data:\n", df)

def permutations(args, df=df, support_confidence=0.6):
    total_transactions = len(df)
    for i in range(len(args)):
        for j in range(len(args)):
            if i == j:
                continue
            item_i = args[i]
            item_j = args[j]
            
            count = 0
            for items in df["Items"]:
                if item_i in items and item_j in items:
                    count += 1
            
            support = count / total_transactions
            
            # Only print pairs above support threshold
            if support >= support_confidence:
                print(f"{item_i} => {item_j} | Count: {count}, Support: {support:.2f}")

            

def Appori(df=df,support_confidence=0.8):

    #find unique items
    temp_1=(df['Items'].tolist())
    flattened_list = [item for sublist in temp_1 for item in sublist]
    unique_items=set(flattened_list)


    #initiate a new df2 which holds values of count and support
    df2=pd.DataFrame(
        {'Item':list(unique_items)}
    )

    item_counts = pd.Series(flattened_list).value_counts()
    total_transaction=df['Items'].count()


    #Now initialize support


    df2['Count'] = df2['Item'].map(item_counts)
    df2['Support']=df2['Count']/total_transaction
    print("Total:",total_transaction)
    
    df2=df2[df2['Support']>=support_confidence]
    #Assuming that we don't have to take out any transaction
    
    # Now to count A->B, B->A for all possibilities
    print(df2)
    
    print("perm")
    permutations(df2['Item'].to_list(),df=df)



Appori(df,support_confidence=0.8)

```
