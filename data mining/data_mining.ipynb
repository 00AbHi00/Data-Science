## Data Mining

Data mining is the process of discovering patterns, correlations, anomalities, and significance from large  data sets using Stats, machine learning and database techniques

<br/>

Examples of Data mining
- Detecting fraud 
- Movie recommendation
- Time series analysis

<br/>

1. Data Cleaning- Handle missing data
2. Data Integration- Combine multiple dataset
3. Data selection- Selecting relvenet data
4. Data Transformation- Normalize or scale values
5. Data Mining- Apply algorithms **Clustering** , **Classification**, **Regression**
6.  Evaluation- Detect patterns
7.  Visualization- Use plots, charts and dashboard
